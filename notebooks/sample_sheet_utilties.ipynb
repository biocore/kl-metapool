# Sample Sheet Utilities

```python
from ipyfilechooser import FileChooser
from metapool.sample_sheet import load_sample_sheet
```

## Load & Validate a Sample Sheet

Use the file chooser below to select a sample sheet file.

```python
starting_dir = "/Users/you/Desktop"
```

```python
fc = FileChooser(starting_dir)
fc._gb.layout.display = None
fc._cancel.layout.display = None
fc._filename.placeholder = 'selected file'
display(fc)
```

Now attempt to load the selected file into a valid samplesheet object:

```python
if fc.selected is None:
    print("No file selected. Please select a file above.")
else:
    try:
        sheet = load_sample_sheet(fc.selected)
        print(f"\nSuccess! Loaded sample sheet with header info:\n {sheet._HEADER}")
    except Exception as e:
        print(f"\nError loading sample sheet:\n: {e}")
```
